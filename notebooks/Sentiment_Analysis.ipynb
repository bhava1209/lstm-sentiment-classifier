## 3. Real-Time Inference on Custom Sentences

```python
predictor = SentimentPredictor(model_path='../models/sentiment_lstm.keras', tokenizer_path='../models/tokenizer.pickle')

test_sentences = [
    'The product quality is excellent and I am very happy.',
    'The product is very poor and I am disappointed.',
    'Fast shipping and wonderful build quality!',
    'Awful customer service, broke after two hours.'
]

for s in test_sentences:
    res = predictor.predict(s)
    print(f"Sentence:   {res['text']}")
    print(f"Sentiment:  {res['sentiment']} ({res['confidence']:.2f}% confidence)")
    print('-' * 60)
```
